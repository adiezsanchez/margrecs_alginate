# Batch Processing for Multiple LIF Containers (Ki67)

This notebook runs nuclei segmentation and per-nucleus feature extraction in batch mode for all or selected `.lif` containers.

Outputs are written to `../results/<lif_container_id>/<lif_image_name>.csv`.

```python
import sys
from pathlib import Path

import pandas as pd

cwd = Path.cwd().resolve()
src_dir = cwd / "src" if (cwd / "src").exists() else cwd.parent / "src"
if str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

from utils.batch_processing import build_runtime_config, run_batch

pd.set_option("display.max_columns", None)
```

```python
# Paths
RAW_DATA_DIRECTORY = r"Z:\margrethe.c.stahl\Ki-67-proliferation"
RESULTS_ROOT = Path("../results")

# Marker configuration: (marker_name, channel_index)
MARKERS = (
    ("DAPI", 0),
    ("F-Actin", 1),
    ("ki67", 2),
    ("brightfield", 3),
)

# Nuclei segmentation settings
NUCLEI_CHANNEL = 0
NUCLEI_LABELING = "voronoi-otsu"  # "cellpose" or "voronoi-otsu"
SPOT_SIGMA = 2
OUTLINE_SIGMA = 2
MIN_MAX_NUCLEI_VOLUME = None  # e.g. (500, 6500)

# Batch behavior
CONTAINER_INDICES: list[int] | None = None  # e.g. [0, 2, 3]
IMAGE_INDICES: list[int] | None = None      # e.g. [0, 1, 5]
OVERWRITE_CSV = False
```

```python
NOTEBOOK_CONFIG = {
    "results_root": str(RESULTS_ROOT),
    "markers": [
        {"name": name, "channel": channel}
        for name, channel in MARKERS
    ],
    "nuclei_channel": NUCLEI_CHANNEL,
    "nuclei_labeling": NUCLEI_LABELING,
    "spot_sigma": SPOT_SIGMA,
    "outline_sigma": OUTLINE_SIGMA,
    "min_max_nuclei_volume": (
        list(MIN_MAX_NUCLEI_VOLUME)
        if MIN_MAX_NUCLEI_VOLUME is not None
        else None
    ),
    "container_indices": CONTAINER_INDICES,
    "image_indices": IMAGE_INDICES,
    "overwrite_csv": OVERWRITE_CSV,
}

CONFIG = build_runtime_config(
    user_config=NOTEBOOK_CONFIG,
    raw_data_directory=RAW_DATA_DIRECTORY,
)
CONFIG
```

```python
batch_results = run_batch(CONFIG)
batch_results
```
